# Text preprocessing 2 Continuation (Convert text to vectors)

#### Bag of Words (BoW) 
A simple way to convert text to vectors by counting the frequency of each word in a document. It creates a vector where each element represents the count of a specific word in the document.

#### TF-IDF (Term Frequency-Inverse Document Frequency)
TF-IDF is a more advanced technique that not only counts the frequency of words but also considers how important a word is in a document relative to a collection of documents. It gives higher weight to words that are unique to a document and lower weight to common words.

Fomula:

    Term Frequency TF = (Number of times a word appears in a document) / (Total number of words in the document)

    IDF = log(Total number of documents / Number of documents containing the word)

    TF-IDF = TF * IDF


Eg:

        Doc 1: He is a good boy
        Doc 2: She is a good girl
        Doc 3: Boy and girl are good

After stop words:

        Doc 1: good boy
        Doc 2: good girl
        Doc 3: boy girl good

#### Term Frequency -> Computed for sentence

                Doc 1      Doc 2      Doc 3
        good    1/2         1/2         1/3

        boy     1/2         0            1/3

        girl    0           1/2         1/3


#### Inverse document Frequency -> Computed for Words

        Words               IDF

        good               log(3/3)=0

        boy                log(3/2)

        girl               log(3/2)


#### TF - IDF

                f1              f2                     f3
                good            boy                     girl

        Doc 1       0               (1/2)*log(3/2)          0

        Doc 2       0               0                       (1/2)log(3/2)

        Doc 3       0               (1/3)log(3/2)           (1/3)log(3/2)




Adv:
1) Intensive
2) Word importance is getting captured

DisAdv:
1) Sparsity
2) Out of Vocabuary

```python
### Get the text data
import os
BASE_DIR = os.getcwd()#path.dirname(os.path.abspath("__vsc_ipynb_file_"))
WIKI_DATA = os.path.join(BASE_DIR, 'data_set', 'wiki_data.txt')
with open(WIKI_DATA, "r") as f:
    content = f.read() 
```

```python
import nltk
```

```python
# Tokenization - Converting para in sentence
nltk.download("punkt")
sentences=nltk.sent_tokenize(content)
```

```python
print(len(sentences))
print(sentences[0])
```

```python
# stemming and stopwords libs
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
stemmer=PorterStemmer()
stemmer.stem("history")
```

```python
# lemmatization
nltk.download("wordnet")
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
lemmatizer.lemmatize("history")
```

```python
import re
corpus = []
for sentence in sentences:
    docs = re.sub('[^a-zA-Z]', ' ', sentence)
    docs = docs.lower()
    corpus.append(docs)
```

```python
corpus
```

```python
## English stop words
nltk.download("stopwords")
stopwords_en=stopwords.words("english")
```

```python
## Stem the words in corpus
for docs in corpus:
    words = nltk.word_tokenize(docs) # Converting sentence(docs) into words
    for word in words:
        if word not in set(stopwords_en):
            print(stemmer.stem(word))
```

```python
#lemmatize the words in the corpus
for docs in corpus:
    words = nltk.word_tokenize(docs) # Converting sentence(docs) into words
    for word in words:
        if word not in set(stopwords_en):
            print(lemmatizer.lemmatize(word))
```

```python
def apply_lemmatization(docs:str):
    """Apply stop words and lemmatization"""
    words = nltk.word_tokenize(docs)
    words = list(map(lambda x:lemmatizer.lemmatize(x) if x not in stopwords_en else '', words))
    clean_docs = " ".join(words)
    return clean_docs

clean_corpus = []
for docs in corpus:
    clean_docs = apply_lemmatization(docs=docs)
    clean_corpus.append(clean_docs)
```

```python

### Bag of Words with n grams
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(binary=True, ngram_range=(3,3))
X=cv.fit_transform(clean_corpus)
cv.vocabulary_
```

```python
X[0].toarray()
```

```python
### TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer
tv = TfidfVectorizer()
X = tv.fit_transform(clean_corpus)
tv.vocabulary_
```
