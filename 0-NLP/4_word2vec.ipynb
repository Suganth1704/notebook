#### Word Embedding:

It is a technique which conversts words into vectors.

##### Types of Word Embedding:
* Count of Frequency - BoW, tf-idf, ONE
* Deep learning trained model - Word2vEC -> CBoW, Skip Grams

#### Word2Vec

* Limited dimension vector representation of words.
* Sparsity is reduced.
* Simantic meaning is preserved.

#### Cosine Similarity
* Similarity between words can be calculated using cosine similarity which is given by the formula:

        Cosine Similarity = (A.B) / (||A|| * ||B||)  
            or
        Cosine Similarity = 1-costeta(A, B)
* Value towards 0 is more similar and value towards 1 is less similar. 


#### CBoW

1) Choose the window size(helps to create the training data) 
2) Traing data - > Independent feature and o/p feature.
3) Text to vector using BoW

Now , Independent feature, o/p feature and vector represtation is given to ANN(fully connected layer)
* Input layer is 4 -independent feature
* Hiden layer - Window size = 5
* ouput layer is softmax layer - 7 nodes in the o/p as the vector dimetion is 7
