##### Word2Vec rewind
* word2Vec is of two types - CBow AND skip gram.
* Using gensim we can create word2vec model.
* Using gensim, we can use the pretrain model or we can train our on model.
* In word2vec we converts word to fixed size of vectors.

For Eg:
* In word2vec we convert words to fix dimentions.
* Lets take a word King and queen we convert these both words to same fixed dimentions.
 [king]300x1 [Queen]300x1.
* This is for two words what if we have more words, obiously we will have more words and we will have more vectors.
[king]300x1 [Queen]300x1 ........ 
* we will have vectors for all the words in our corpus.
* To over come this problem we can use Average word2vec.
* In average word2vec we take the average of all the vectors of the words in a sentence and we get a single vector for the whole sentence.
