##### RNN - Recurrent nural network
- RNN are a class of nureal networks that are designed to handle sequential data by retaining the information from the previous steps.
- They are sepecially effective for tasks where context and order matter.




##### Types of RNN:
1) <b>One to One RNN</b> - One input one output, Eg: Image classification
2) <b>One to Many RNN</b> - One input Many output, Eg: Music generation, text generation, Google search suggestion, Movie suggestion.
3) <b>Many to One RNN</b> - Many input one output, Eg: Sentiment analysis.
4) <b>Many to Many RNN</b> - Many input Many output, Eg: Language tanslation, chat bots, Question and answering.

#### Backward propagation in RNN:

- For Backward propagation like forward propagation we'll get the loss (y-y^).
- Then we propagate the loss back through the network to update the weights and bias in each timestamp.
- Let say the RNN is so deep and the timestamps are long so while updating the wights in the backward at some poin the value will be negligible and it will not update the weights and bias and this is called <b>vanishing gradient problem</b>.



- vanishing gradient problem - For example if the nuron is using an  activation fuction like sigmoid. Sigmoid function ranges between 0 and 1. So while updatin the wights the derivate of sihmoid function will rnage between 0 and 0.25. So if the wights are updated with the derivate of sigmoid function then at somepointthe wights will be updated with a very small value and it will not update the weights and bias and this is called vanishing gradient problem.
- To solve this problem we have LSTM RNN.
