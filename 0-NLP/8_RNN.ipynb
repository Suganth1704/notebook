##### RNN - Recurrent nural network
- RNN are a class of nureal networks that are designed to handle sequential data by retaining the information from the previous steps.
- They are sepecially effective for tasks where context and order matter.




##### Types of RNN:
1) <b>One to One RNN</b> - One input one output, Eg: Image classification
2) <b>One to Many RNN</b> - One input Many output, Eg: Music generation, text generation, Google search suggestion, Movie suggestion.
3) <b>Many to One RNN</b> - Many input one output, Eg: Sentiment analysis.
4) <b>Many to Many RNN</b> - Many input Many output, Eg: Language tanslation, chat bots, Question and answering.
