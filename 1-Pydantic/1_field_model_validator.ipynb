#### Field validator

```python
from pydantic import BaseModel, EmailStr, Field, AnyUrl, field_validator#, model_validator, computed_field

class Student(BaseModel):
    name: str
    age: int=Field(gt=0, le=100)
    linked_in: AnyUrl
    college_email: EmailStr

    @field_validator('college_email')
    @classmethod
    def email_validator(cls, value):
        domain = value.split("@")[1]
        if domain != "abc.com":
            raise ValueError('Email is not college email')
        return value

student = {'name':'Suganth',
                'age': 20,
                'linked_in': 'https://www.linkedin.com/',
                'college_email': 'suagnth@abc.com'}

student_data = Student(**student)
```

```python
print(student_data.name)
print(student_data.age)
print(student_data.linked_in)
print(student_data.college_email)
```

#### Model validator - To validate one or more fields using single method

```python
from pydantic import BaseModel, EmailStr, Field, AnyUrl, model_validator

class NotEligibleForCollege(Exception):
    pass

class Student(BaseModel):
    name: str
    age: int=Field(gt=0, le=100)
    school_completed: bool
    linked_in: AnyUrl
    college_email: EmailStr

    @model_validator(mode='after')
    @classmethod
    def model_validator(cls, model):
        if not model.school_completed or model.age < 18:
            raise NotEligibleForCollege('Not Eligible for the college admission')
        return model

student = {'name':'Suganth',
                'age': 20,
                'school_completed': True,
                'linked_in': 'https://www.linkedin.com/',
                'college_email': 'suagnth@abc.com'}

student_data = Student(**student)
```

```python
print(student_data.name)
print(student_data.age)
print(student_data.linked_in)
print(student_data.college_email)
print(student_data.school_completed)
```

```python
from pydantic import BaseModel, EmailStr, Field, AnyUrl, computed_field

class Student(BaseModel):
    name: str
    age: int=Field(gt=0, le=100)
    school_completed: bool
    linked_in: AnyUrl
    college_email: EmailStr

    @computed_field
    @property
    def twice_age(self) -> int:
        return 2* self.age

student = {'name':'Suganth',
                'age': 20,
                'school_completed': True,
                'linked_in': 'https://www.linkedin.com/',
                'college_email': 'suagnth@abc.com'}

student_data = Student(**student)
```

```python
print(student_data.name)
print(student_data.age)
print(student_data.linked_in)
print(student_data.college_email)
print(student_data.school_completed)
print(student_data.twice_age)
```

#### Nested Fields

```python
class Address(BaseModel):
    house_num: int
    street: str

class Student(BaseModel):
    name: str
    age: int=Field(gt=0, le=100)
    school_completed: bool
    linked_in: AnyUrl
    college_email: EmailStr
    address: Address

address = {'house_num':1, 'street': 'oak street'}

student = {'name':'Suganth',
                'age': 20,
                'school_completed': True,
                'linked_in': 'https://www.linkedin.com/',
                'college_email': 'suagnth@abc.com',
                'address': Address(**address)}

student_data = Student(**student)
```

```python
print(student_data.name)
print(student_data.age)
print(student_data.linked_in)
print(student_data.college_email)
print(student_data.school_completed)
print(student_data.address)
```
