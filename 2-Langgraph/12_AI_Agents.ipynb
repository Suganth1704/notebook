#### What is an AI Agent?

An AI Agent is an agent with LLM as brain whcih can think, decide and take aactions to aive a goal.

Insted of just answering a prompt, it can

* reason
* call tools
* make decisions
* remember previous steps
* retry when something fails

Example

    -> User: Book me the cheapest flight to Bangalore

            Understand request

                ↓

            Search flights

                ↓

            Compare prices

                ↓

            Ask for missing information

                ↓

            Book ticket

                ↓

            Send confirmation

#### What is ReAct?

ReAct = Reason + Act

The ReAct agent are the agents that alters between reasoning about the problem, taking action, observing the results and repaeating util it reaches a final answer.

```python
import os
from dotenv import load_dotenv
load_dotenv()
from langchain.agents import create_agent

##llm
from langchain_groq import ChatGroq
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
llm = ChatGroq(model="openai/gpt-oss-120b")
```

```python
from mcp import ClientSession
from mcp.client.streamable_http import streamablehttp_client
from langchain_mcp_adapters.tools import load_mcp_tools


async with streamablehttp_client(
        "http://127.0.0.1:8000/mcp"
    ) as (read, write, _):
    async with ClientSession(read, write) as session:
        await session.initialize()

        tools = await session.list_tools()
        print(tools)
        tool_list = await load_mcp_tools(session=session)
        # result = await session.call_tool("add" ,{"expression":"2+2"})
        agent = create_agent(
            model=llm,
            tools=tool_list,
            system_prompt='Check for tools first before each action'
        )
        resp = await agent.ainvoke(
            {"messages": [{"role":"user", "content":"Check the weather in chennai and translate response in Tamil"}]}
        )
for msg in resp['messages']:
    msg.pretty_print()
```
