#### Prompt engineering:

- Prompt engineering is desing and refining instructions. examples, constraints, and output requiremetns to guide an LLM towards the desired response.

Eg:

        # ROLE
        You are a customer-support agent.

        # OBJECTIVE
        Resolve the customer's issue using the available tools.

        # RULES
        1. Never invent order information.
        2. Use get_order_details before discussing an order.
        3. If the requested action requires authorization, ask for confirmation.
        4. If you cannot resolve the issue, escalate it.

        # TOOL USAGE
        Use tools only when their information is required.

        # OUTPUT
        Return:
        {
            "resolution": "...",
            "action_taken": "...",
            "needs_escalation": true/false
        }

        # USER
        {user_message}

        # RELEVANT CONTEXT
        {retrieved_context}

#### Context Engineering:

- Context engineering is designing and dynamically managing all the informations available to the LLM -- including instructions, user input , conversation, conversation history, memeory, retrieved knowledge, tool definitionsm tool results, and agent state so the model receives the right information, in the right format at right time.

"What should the model know and see right now to successfully perform this task?"

    Context Engineering
    │
    ├── Prompt / Instructions
    ├── User input
    ├── Conversation history
    ├── Memory
    ├── RAG results
    ├── Tool definitions
    ├── Tool results
    ├── Agent state
    ├── Examples
    └── Output constraints

#### Harness Engineering

- Harness engineering is the engineering of the environment around an AI model/agent so that it can reliably accomplish a rask in a real system.

                        │    AI Agent         │

            ↓                      ↓                      ↓
         Tools                  Memory                 RAG
            ↓                      ↓                      ↓
         APIs                  Database              Vector DB
            │                      │                      │
            └──────────────────────┼──────────────────────┘
                                    ↓
                           Agent Runtime
                                    │
                     ┌──────────────┼──────────────┐
                     ↓              ↓              ↓
                  Logging        Tracing        Evaluation
                     ↓              ↓              ↓
                  Guardrails     Retries       Monitoring
                                    │
                                    ↓
                              Human approval

The harness handles things like:

- tool execution                
- permissions           
- authentication            
- retries           
- timeouts          
- state management          
- human-in-the-loop         
- error handling
- validation
- logging
- tracing
- evaluation
- cost limits
- token limits
- sandboxing
- observability
- recovery from failures            

LLM = engine

Prompt = instructions to the driver

Context = information available to the driver

Harness = the entire vehicle + controls + safety systems

#### Loop engineering:

- It is the practive of desinging the agent's iterative execution loop, how an agent repeatedly observes, decides, acts, checks the results, and decides what to do next.

##### Basic agent loop:

        User Request
            ↓
        Observe
            ↓
        Decide
            ↓
        Act ──────→ Tool
            ↓           │
        Observe ←─────┘
            ↓
        Check Result
            │
            ├── Not done → Decide → Act → ...
            │
            └── Done → Final Answer


Eg:

        Generate
        ↓
        Evaluate
        ↓
        Good?
        ┌─┴─┐
        Yes  No
        ↓    ↓
      Done  Improve
            ↓
            Generate
