#### REST APIs [Source: https://realpython.com/api-integration-in-python/#rest-architecture]
* REST - Representational state transfer is a software architecture style that defines a pattern for client and server communication over a network.

1) Stateless: The server won’t maintain any state between requests from the client.
2) Client-server: The client and server must be decoupled from each other, allowing each to develop independently.
3) Cacheable: The data retrieved from the server should be cacheable either by the client or by the server.


#### HTTP Methods
    GET - Retrive an existing resource.
    POST - Create a new resource.
    PUT - Update an existing resource.
    PATCH - Partically update an existing resource.
    DELETE - Delete a resource.

```python
# GET
import requests
api_url = "https://jsonplaceholder.typicode.com/todos/1"
response = requests.get(api_url)
```

```python
response.json()
```

#### HTTP Headers
* A dict contains metadata about the response, such as Content-Type of the reponse.
* HTTP headers let the client and the server pass additional information with a message in a request or response. 


```python
response.headers
```

```python
# POST
api_url = "https://jsonplaceholder.typicode.com/todos/"
todo = {"userId": 1, "title": "Buy milk", "completed": False}
response = requests.post(api_url, json=todo)
```

```python
response.status_code
```

```python
response.json()
```

* When you do this, requests.post() automatically sets the request’s HTTP header Content-Type to application/json. It also serializes todo into a JSON string, which it appends to the body of the request.
* If you don’t use the json keyword argument to supply the JSON data, then you need to set Content-Type accordingly and serialize the JSON manually. Here’s an equivalent version to the previous code:

```python
import json
header = {"Content-type":"application/json"}
response = requests.post(api_url, data=json.dumps(todo), headers=header)
```

```python
print(response.text)
```

```python
# PUT
api_url = "https://jsonplaceholder.typicode.com/todos/10"
response = requests.get(api_url)
response.json()
```

```python
todo = {"userId": 1, "title": "Wash car", "completed": True}
response = requests.put(api_url, json=todo)
```

```python
response.json()
```

```python
todo = {"title": "Car Wash"} 
response = requests.patch(api_url, json=todo)
print(response.status_code)
print(response.json())
```

#### PUT vs PATCH
* PUT updates all fields on the resource with new data.
* PATCH requests only update a part of a resource. In the request

```python
# DELETE
response = requests.delete(api_url)
response.json()
```

```python
response.status_code
```

#### Design Success Responses

HTTP request:

    GET /cars HTTP/1.1
    Host: api.example.com

    HTTP request made of four parts:
    1) HTTP Method - GET, POST, PUT, PATCH, DELETE
    2) Api end point - /cars
    3) HTTP version - HTTP/1.1
    4) API Host - api.example.com

GET:
* Request:
    - A GET request should never modify an existing resource. If the request contains data, then this data should be ignored and the API should return the resource unchanged.
* Response:
    - It’s important to always set the correct Content-Type header on your response. If you send JSON, then set Content-Type to application/json. If XML, then set it to application/xml. This header tells the user how they should parse the data.
    - For any successful GET request, you should return 200 OK

            HTTP/1.1 200 OK
            Content-Type: application/json

            {
                "id": 1,
                "make": "GMC",
                "model": "1500 Club Coupe",
                "year": 1998,
                "vin": "1D7RV1GTXAS806941",
                "color": "Red"
            },

POST:
* Request:
    - POST request should include  new record.
    - Content-Type

            POST /cars HTTP/1.1
            Host: api.example.com
            Content-Type: application/json

            {
                "make": "Nissan",
                "model": "240SX",
                "year": 1994,
                "vin": "1N6AD0CU5AC961553",
                "color": "Violet"
            }
* Response:
    - Reesponse should have "201 Created" status code to tell the user that a new resource was created
    - It’s important to always send back a copy of a resource when a user creates it with POST or modifies it with PUT or PATCH. This way, the user can see the changes that they’ve made.

            HTTP/1.1 201 Created
            Content-Type: application/json

            {
                "id": 4,
                "make": "Nissan",
                "model": "240SX",
                "year": 1994,
                "vin": "1N6AD0CU5AC961553",
                "color": "Violet"
            }

PUT:
* Reqeuest:
    - Request should have all the fields of the record.
    - PUT updates all fields on the resource with new data

            PUT /cars/4 HTTP/1.1
            Host: api.example.com
            Content-Type: application/json

            {
                "make": "Buick",
                "model": "Lucerne",
                "year": 2006,
                "vin": "4T1BF3EK8AU335094",
                "color":"Maroon"
            }

* Response:
    - Status code should be 200 ok
    -  Again, you always want to send back the full resource for a PUT request.

            HTTP/1.1 200 OK
            Content-Type: application/json

            {
                "id": 4,
                "make": "Buick",
                "model": "Lucerne",
                "year": 2006,
                "vin": "4T1BF3EK8AU335094",
                "color":"Maroon"
            }

PATCH:
* Request:
    - PATCH requests only update a part of a resource. In the request below, the vin and color fields will be updated with new values.

            PATCH /cars/4 HTTP/1.1
            Host: api.example.com
            Content-Type: application/json

            {
                "vin": "VNKKTUD32FA050307",
                "color": "Green"
            }
* Response:
    - The response should have the full copy of the car.

            HTTP/1.1 200 OK
            Content-Type: application/json

            {
                "id": 4,
                "make": "Buick",
                "model": "Lucerne",
                "year": 2006,
                "vin": "VNKKTUD32FA050307",
                "color": "Green"
            }


DELETE:
* Request:
    - DELETE request tells the API to remove the resource with the particular identifier.

            DELETE /cars/4 HTTP/1.1
* Response:
    - Response only includes the status code 204 No Content
    - Status code tells a user that the operation was successful, but no content was returned in the response.

            HTTP/1.1 204 No Content
