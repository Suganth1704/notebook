#### Design Error Responses

404 - Not Found:
* Request:
        - Below is the request which is not avilable in the resource.
            
            GET /motorcycles HTTP/1.1
            Host: api.example.com
* Response:
    - If the response was not found, API should send back the following response.
    - Response includes a 404 Not Found status code. Along with this, the response contains a JSON object with a descriptive error message.

            HTTP/1.1 404 Not Found
            Content-Type: application/json
            ...

            {
                "error": "The requested resource was not found."
            }
400 - Bad Request
* Request: 
    - The POST request contains JSON, but it isn’t formatted correctly. It’s missing a closing curly brace (}) at the end. The API won’t be able to process this data.

            POST /cars HTTP/1.1
            Host: api.example.com
            Content-Type: application/json

            {
                "make": "Nissan",
                "year": 1994,
                "color": "Violet"
* Response:
    - API should respond with 400 Bad request status code, telling the user they need to fix the request.

            HTTP/1.1 400 Bad Request
            Content-Type: application/json

            {
                "error": "This request was not properly formatted. Please send again."
            }


415 - Unsupported Media Type
* Sending xml codentent insted of JSON in POST request.
422 - Unprocessable Entity
* Request: 
    - In the request the user might have added the fileds which were not supported by the API.

            POST /cars HTTP/1.1
            Host: api.example.com
            Content-Type: application/json

            {
                "make": "Nissan",
                "model": "240SX",
                "topSpeed": 120 -> invalid data
                "warrantyLength": 10 -> invalid data
            }

* Response:
    - The Response includes the 422 Unprocessable Entity status code.

            HTTP/1.1 422 Unprocessable Entity
            Content-Type: application/json

            {
                "error": "Request had invalid or missing data."
            }
