#### Rate limiters:
* A rate limiter in API development is a mechanism that controls how may requests a client can make to an APU within a specific time window, ensure fair usage and protecting servers from overload or abuse.
* It's essential for preventing denial-of -service (DoS) attacks, brute-force attempts, and excessive traffic that could degrade performance.

* Why rate limiters:
    - Prevents server overload
    - Ensures fair usage
    - Protects against abuse
    - Improves reliability
