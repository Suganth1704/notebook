##### What is SFT?

* SFT - Supervised Fine-Tuning
* Why SFT - Every example has the correct answer.

Eg: 

        > Input:

        ↓

        > Target:

        Run spark-submit again after verifying...

            Restart your laptop. [Wrong answer]

            Prediction vs Expected Output
        
        > Then gradients are computed.
        > Weights are updated.
        > Repeat thousands of times.
        > This is exactly like training any neural network     

##### Why LoRA [Low Rank Adaptation] of LLM?

What does it mean?

* Low-Rank: Instead of updating the entire weight matrix of a neural network, LoRA approximates the update using two much smaller matrices with a low rank.
* Adaptation: These small matrices adapt a pre-trained model to a new task without modifying the original model weights.

* Imagine a 7B model.
* Each parameter is updated during training.
* That's huge.
* Needs lots of GPU memory.
* Instead LoRA says

        Freeze everything.

                ↓

        Add tiny matrices.

                ↓

        Train ONLY them.

Instead of training 7,000,000,000 parameters, we train 4,000,000 parameters. That's often less than 1% of the model.

* LoRA doesn't add arbitrary "new parameters" to replace the model. Instead, it adds small trainable adapter matrices to selected layers (typically linear layers like q_proj and v_proj) while keeping the original model weights frozen.

##### Where are these adapters added?

* Inside the Transformer layers.

* Each layer contains linear projections like:

    Attention

    Q

    K

    V

    O

LoRA inserts trainable matrices into these linear layers.

##### Doubt? -> If LoRA freezes the original model, how does the model actually learn?

Imagine a linear layer computes:

        Y=WX

where:

* W is the pretrained weight matrix (frozen)
* X is the input

LoRA modifies this to:

        Y=(W+BA)X

where:

* W remains frozen
* A and B are small trainable matrices (the LoRA adapters)


##### If the model has 7 billion parameters and LoRA only trains 10 million parameters, how can those 10 million parameters significantly change the model's behavior?

Imagine a huge company with 7,000 employees.

You don't need to retrain every employee. If you update the instructions given to 100 team leaders, the behavior of the entire company changes because those leaders influence everyone downstream.

LoRA works similarly --> 

The original weights (W) already contain all the language knowledge. LoRA doesn't replace that knowledge. Instead, it learns small adjustments to the computations in important layers, nudging the model toward the desired behavior.

Note:

- If the base model has zero knowledge about a topic, LoRA cannot magically teach it everything with a tiny adapter. LoRA works best when the base model already has a solid foundation and only needs to be specialized.
