#### Encoder Vs Decoder

- Encoder = understands the input
- Decoder = generates the output

A Transformer can be structured like:

                INPUT
                  │
                  ▼
            ┌───────────┐
            │  ENCODER  │
            └───────────┘
                  │
        Understanding / representation
                  │
                  ▼
            ┌───────────┐
            │  DECODER  │
            └───────────┘
                  │
                  ▼
               OUTPUT

#### Three major Transformer architectures

| Architecture    | Main job                           | Examples                 |
| --------------- | ---------------------------------- | ------------------------ |
| Encoder-only    | Understand/analyze text            | BERT                     |
| Decoder-only    | Generate text                      | GPT, Llama               |
| Encoder-Decoder | Understand input + generate output | T5, original Transformer |


#### 1) Encoder

- Encoder takes the entire input and creates a rich representation of it.
- It can look at the entire sentence
- Eg: 

    The animal didn't cross the road because it was tired.

- When process "it", the encoder can look at 

    The animal didn't cross the road because it was tired
                                              ↑
                                             all

- It uses <b>bidirectional attenttion</b>. So each token can attend to tokens before and after it.

##### --> Encoder attention:

- Suppose: I love cats.

                I     love    cats
        I      0.5    0.3     0.2
        love   0.2    0.5     0.3
        cats   0.1    0.3     0.6

- This means each token can consider all other tokens. This is called self-attention.

#### --> Encoder is usful for

Because the encoder is good at understanding, encoder-only models are commonly used

- Classification
- Sentimental analysis
- Sematic similarity
- Embeddings
- Search

A famous encoder-only model is: <b>BERT</b>

#### 2) Decoder

- The decoder is designed to generate text one token at a time.

Suppose we ask: Whar is the capital of France?

- Decoder generates

        The
        ↓
        The capital
        ↓
        The capital of
        ↓
        The capital of France
        ↓
        The capital of France is
        ↓
        The capital of France is Paris

#### --> Masking

- When predicting "France", it cannot see the future tokens:

    The capital of France is Paris
                ↑          ↑
            allowed     NOT allowed

- The decoder can only attend to previous tokens. This is called Causal / Masked Self-Attention

#### --> Decoder attention matrix

- Prediction: I love cats.
- The decoder's attention might look conceptually like:

                I    love   cats
        I       ✓     ✗      ✗
        love    ✓     ✓      ✗
        cats    ✓     ✓      ✓

- Because when predicting "love", the model shouldn't already know "cats".

Note: For modern agentic systems, the LLM that acts as the "brain" is usually a decoder-only Transformer.

#### 3) Encoder-Decoder

- The decoder in Encoder-Decoder has TWO attentions
- 1) Masked self-attention
- 2) Cross-attention


                        Decoder
                        │
                        ├── Masked Self-Attention
                        │       ↓
                        │   Previous output
                        │
                        └── Cross-Attention
                                ↓
                        Encoder output


| Feature                   | Encoder       | Decoder       | Encoder-Decoder       |
| ------------------------- | ------------- | ------------- | --------------------- |
| Main purpose              | Understand    | Generate      | Understand + generate |
| Attention                 | Bidirectional | Causal/masked | Both                  |
| Sees future input tokens? | Yes           | No            | Encoder: Yes          |
| Generates text?           | Usually no    | Yes           | Yes                   |
| Example                   | BERT          | GPT/Llama     | T5                    |
| Classification            | Excellent     | Possible      | Possible              |
| Text generation           | Not primary   | Excellent     | Excellent             |
| Translation               | Not by itself | Possible      | Excellent             |
