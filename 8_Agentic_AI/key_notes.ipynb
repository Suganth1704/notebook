### AI Agents latency:

1) How to reduce latency?

- Trace every steps
- Chach repeated outouts
- Run independent tasks parallely.
- reduce unnecessary sequences

2) Bottlenecks

- LLM Inference
- Large input and output tokens
- RAG retrival and reranking
- Database quries.

3) What trade off would you make b/w  Latency vs Quality

- Use model routing based on the task complexity
- Reduce reasoning depth
- Increase the context when accuracy matters.

4) Handling production failures

- Log every output
- Use fallback modles
- Set timeouts
- Resume from checkpoints for long running agents.
