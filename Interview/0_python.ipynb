##### Decorators
* My understanding :  Decorators are higer order functions which takes functions as a agrument and modifies it behaviour.
* A decorator is a function that takes another function as input and returns a new function.

```python
### Decorators in python
def decorator(func):
    def wrapper(*args,**kargs):
        funcOut = f"{func(*args,**kargs)} World!!"
        return funcOut
    return wrapper

def dotwice(func):
    def wrapper(*args, **kwargs):
        funcOut = func(*args, **kwargs)
        funcOut += f" {func(*args, **kwargs)}"
        return funcOut 
    return wrapper

@dotwice
@decorator
def sayHello(saidBy:str):
    return f'{saidBy} :Hello'

greet=sayHello(saidBy="Suganth")
```

```python
greet
```

```python
sayHello.__name__
```

* above example shows that the function looses it identity when it use decorator, to prevent the identity we use @functools.wrap(func)

```python
### Decorators in python - Identity
import functools
def decorator(func):
    @functools.wraps(func)
    def wrapper(*args,**kargs):
        funcOut = f"{func(*args,**kargs)} World!!"
        return funcOut
    return wrapper

def dotwice(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        funcOut = func(*args, **kwargs)
        funcOut += f" {func(*args, **kwargs)}"
        return funcOut 
    return wrapper

def repeat(n=2):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kargs):
            funcOut =""
            for _ in range(n):
                funcOut += f" ,{func(*args,**kargs)} World!! from repeat"
            return funcOut
        return wrapper
    return decorator



@dotwice
@decorator
def sayHello(saidBy:str):
    return f'{saidBy} :Hello'


@repeat(n=4)
def sayHelloR(saidBy:str):
    return f'{saidBy} :Hello'

greet=sayHello(saidBy="Suganth")
greetR=sayHelloR(saidBy="Suganth")
```

#### Class as decorator
* Point to remeber, func = decorator(func)

```python
class Counter:
    def __init__(self, start=0):
        self.count = start
    def __call__(self):
        self.count += 1
        print(f"Current count is {self.count}")
##The .__call__() method is executed each time you try to call an instance of the class:

class Countcall:
    def __init__(self, func):
        self.func = func
        self.nCalls = 0

    def __call__(self, *args, **kwargs):
        self.nCalls +=1
        print(f"Call {self.nCalls} of {self.func.__name__}()")
        return self.func(*args, **kwargs)
```

```python
@Countcall
def sayHelloC(saidBy:str):
    return f'{saidBy} :Hello'

sayHelloC(saidBy='Suganth')
sayHelloC(saidBy='Suganth')
```

#### What Is an Iterator?
An iterator is any object that implements
* __iter__() → returns the iterator object
* __next__() → returns the next value or raises StopIteration

Key Properties:
* Returns one item at a time and uses lazy evaluation (doesn't load everthing into memory)
* Created using iter()

#### Generator
A generator is:
* A function that uses yield instead of return
* Automatically creates an iterator object
* Pauses and resumes execution between yields


```python
def gen(li:list):
    for i in li:
        yield i

li = [1,2,3,4,5]
op=gen(li)
```

```python
for i in op:
    print(i)
```

The key idea is:

* self refers to the current object (instance).
* cls refers to the class itself.
* @classmethod is a decorator that tells Python to pass the class (cls) as the first argument instead of an instance.

```python
## One of the biggest uses of 
# @classmethod is creating alternative constructors.

class Employee:

    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    @classmethod
    def from_string(cls, data):
        name, age = data.split("-")
        return cls(name, int(age))
```

```python
emp = Employee.from_string("Alice-25")
print(emp.name)
print(emp.age)
```

The purpose of @property is to let you access a method like an attribute, while still allowing you to add validation or computation behind the scenes.

```python
class Employee:

    def __init__(self,name, salary):
        self.name = name
        self.salary = salary

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, value):
        if value < 0:
            raise ValueError("Salary cannot be -ve")
        self._salary = value

    #read only
    @property
    def bonuns(self):
        return self.salary * 0.15
```

```python
emp = Employee(name="Suganth", salary=35000)
print(emp.bonuns) # read only property
emp.salary = 40000
print(emp.bonuns)
emp.bonuns = 10000
```

##### Multi-threading in python:

A thread is the smallest unit of execution within a process. Threads in the same process share memory and resources.

2. What is the GIL?

The Global Interpreter Lock is a mechanism in CPython that allows only one thread to execute Python bytecode at a time, simplifying memory management.

3. Does multithreading make Python faster?

* CPU-bound tasks: Usually no, because of the GIL.
* I/O-bound tasks: Yes, because threads release the GIL while waiting for I/O.

4. What is a race condition?

A race condition occurs when multiple threads access and modify shared data simultaneously, leading to unpredictable results.

```python
from concurrent.futures import ThreadPoolExecutor

import requests
import json
import time

def make_request(url:str):
    print(url)
    resp = requests.get(url)
    resp_dict = json.loads(resp.text)
    time.sleep(5)
    return resp_dict

url_list = ["https://official-joke-api.appspot.com/jokes/random/5",
"https://official-joke-api.appspot.com/jokes/random/25",
"https://official-joke-api.appspot.com/jokes/random/250",
"https://official-joke-api.appspot.com/jokes/random/51",
"https://official-joke-api.appspot.com/jokes/random/26",
"https://official-joke-api.appspot.com/jokes/random/24"]

with ThreadPoolExecutor(max_workers=4) as executor:
    result = executor.map(make_request, url_list)

for i in result:
    print(i)
```
