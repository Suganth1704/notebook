##### Decorators
* My understanding :  Decorators are higer order functions which takes functions as a agrument and modifies it behaviour.
* A decorator is a function that takes another function as input and returns a new function.

```python
### Decorators in python
def decorator(func):
    def wrapper(*args,**kargs):
        funcOut = f"{func(*args,**kargs)} World!!"
        return funcOut
    return wrapper

def dotwice(func):
    def wrapper(*args, **kwargs):
        funcOut = func(*args, **kwargs)
        funcOut += f" {func(*args, **kwargs)}"
        return funcOut 
    return wrapper

@dotwice
@decorator
def sayHello(saidBy:str):
    return f'{saidBy} :Hello'

greet=sayHello(saidBy="Suganth")
```

```python
greet
```

```python
sayHello.__name__
```

* above example shows that the function looses it identity when it use decorator, to prevent the identity we use @functools.wrap(func)

```python
### Decorators in python - Identity
import functools
def decorator(func):
    @functools.wraps(func)
    def wrapper(*args,**kargs):
        funcOut = f"{func(*args,**kargs)} World!!"
        return funcOut
    return wrapper

def dotwice(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        funcOut = func(*args, **kwargs)
        funcOut += f" {func(*args, **kwargs)}"
        return funcOut 
    return wrapper

def repeat(n=2):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args, **kargs):
            funcOut =""
            for _ in range(n):
                funcOut += f" ,{func(*args,**kargs)} World!! from repeat"
            return funcOut
        return wrapper
    return decorator



@dotwice
@decorator
def sayHello(saidBy:str):
    return f'{saidBy} :Hello'


@repeat(n=4)
def sayHelloR(saidBy:str):
    return f'{saidBy} :Hello'

greet=sayHello(saidBy="Suganth")
greetR=sayHelloR(saidBy="Suganth")
```

#### Class as decorator
* Point to remeber, func = decorator(func)

```python
class Counter:
    def __init__(self, start=0):
        self.count = start
    def __call__(self):
        self.count += 1
        print(f"Current count is {self.count}")
##The .__call__() method is executed each time you try to call an instance of the class:

class Countcall:
    def __init__(self, func):
        self.func = func
        self.nCalls = 0

    def __call__(self, *args, **kwargs):
        self.nCalls +=1
        print(f"Call {self.nCalls} of {self.func.__name__}()")
        return self.func(*args, **kwargs)
```

```python
@Countcall
def sayHelloC(saidBy:str):
    return f'{saidBy} :Hello'

sayHelloC(saidBy='Suganth')
sayHelloC(saidBy='Suganth')
```

#### What Is an Iterator?
An iterator is any object that implements
* __iter__() → returns the iterator object
* __next__() → returns the next value or raises StopIteration

Key Properties:
* Returns one item at a time and uses lazy evaluation (doesn't load everthing into memory)
* Created using iter()

```python
li = [1,2,3,4]
it = iter(li)

print(next(it))
print(next(it))
```

#### Generator
A generator is:
* A function that uses yield instead of return
* Automatically creates an iterator object
* Pauses and resumes execution between yields


```python
def gen(li:list):
    for i in li:
        yield i

li = [1,2,3,4,5]
op=gen(li)
```

```python
for i in op:
    print(i)
```
