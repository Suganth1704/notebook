#### Four pillars, or four tenets of OOP:

* <b>Encapsulation</b> means bundling data and methods together inside a class and controlling access to the object's data.

* <b>Inheritance</b> enables the creation of hierarchical relationships between classes, allowing a subclass to inherit attributes and methods from a parent class. This promotes code reuse and reduces duplication.

* <b>Abstraction</b> focuses on hiding implementation details and exposing only the essential functionality of an object. By enforcing a consistent interface, abstraction simplifies interactions with objects, allowing developers to focus on what an object does rather than how it achieves its functionality.

* <b>Polymorphism</b> allows you to treat objects of different types as instances of the same base type, as long as they implement a common interface or behavior. Python’s duck typing make it especially suited for polymorphism, as it allows you to access attributes and methods on objects without needing to worry about their actual class.


```python
## Singleton class

class Config:
    _instance = None

    def __new__(cls, *args, **kwargs):
        print("In __new__")
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.initalized = False
        print(vars(cls._instance))
        return cls._instance

    def __init__(self, enabled=False):
        print("In __init__")
        if not self.initalized:
            self.enabled=enabled
            self.initalized = True
```

```python
## Encapsulation:
"""
name       → public
_name      → protected convention
__name     → private-like / name mangling
"""

class Student:
    def __init__(self, name, age):
        self.name = name
        self.__age = age

    @property
    def age(self):
        return self.__age

s = Student("Name", 26)
print(s.name)
print(s.age)

vars(s)
```

```python
## Abstraction
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

class CrediCard(Payment):

    def pay(self, amount):
        print(f"Paid {amount} using credi card")

class UPI(Payment):

    def pay(self, amount):
        print(f"Paid {amount} using UPI")
```

```python
## Inheritance

class Animal:

    def __init__(self, typeof):
        self.typeof = typeof

    def eat(self):
        print("Eating")

class Dog(Animal):

    def __init__(self, typeof, name):
        # Animal.__init__(self, typeof) # -> replace with super
        super().__init__(typeof)
        self.name = name

    def bark(self):
        Animal.eat(self) #-> replace with super
        super().eat()
        print("Barking")
```

```python
d= Dog("Domestic", "Deby")
d.bark()
print(d.typeof)
print(d.name)
```

```python
## Polymorphism - same name but behaves differently

class Dog:

    def make_sound(self):
        print("Wow wow")

class Duck:

    def make_sound(self):
        print("Quack quack")

animals = [Dog(), Duck()]
for animal in animals:
    animal.make_sound() # duck typing

# Python special form of poymorphism is duck typing - if it walks like a duck and quacks like
# a duck, treat it like a duck.
```

```python
## Method overriding

class Animal:

    def speak(self):
        print("Animal sound")


class Dog(Animal):

    def speak(self): # overring the parent class method
        print("Bark")
```
