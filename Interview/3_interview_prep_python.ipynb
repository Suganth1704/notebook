```python
## Decorator

## Higher order funtion which takes another function as input and 
# modifies it output and retunrs a new fucntion.
import functools

def double_it(func):
    @functools.wraps(func)
    def wrapper(*args,**kwargs):
        s_li = list(map(lambda x: x**2, func(*args, **kwargs)))
        return s_li
    return wrapper

@double_it
def return_list(li:list) -> list:
    return li

return_list(li=[1,2,3,4,5])
```

```python
## Singleton pattern using decorator

def singleton(cls):
    _instance = {}
    def wrapper(*args,**kwargs):
        if not _instance.get(cls, None):
            _instance[cls] = cls(*args, **kwargs)
        return _instance[cls]
    return wrapper

@singleton
class Student:
    pass

s1 = Student()
s2 = Student()

s1 is s2
```

```python
## Iterator is a python object that implements the iterator 
# protocol using __iter__() and  __next__() and produces value one at a time.

class Counter:

    def __init__(naandhan, max_size): ## self -> naandhan
        naandhan.count = 0
        naandhan.max_size = max_size


    def __iter__(naandhan):
        return naandhan

    def __next__(naandhan):
        if naandhan.count <= naandhan.max_size:
            naandhan.count += 1
            return naandhan.count
        raise StopIteration

ct = Counter(4)

print(next(ct))
print(next(ct))
print(next(ct))
print(next(ct))
print(next(ct))
print(next(ct))
```

```python
## Generators are function based iterator object which yields one value at a time
# which supporte lazy evaluation and helps in managing memeory efficiently

def generator(li:list):
    for i in li:
        yield i

g=generator(list(range(1,5)))

print(next(g))
print(next(g))
print(next(g))
print(next(g))
print(next(g))
print(next(g))
```

```python
## MRO - Method resolution order

# MRO is the order in whihc the python looks for the class when looking for a method or attribute.

class A:
    def show(self):
        print("A")

class B:
    def show(self):
        print("B")

class C(B, A):
    pass

c = C()
c.show()
print(C.__mro__)
```

```python
## __new__ vs __init__

# __new__ creats an empty object for the respective without any attributs
# __init__ initates the objects by binding instance attributs

# __new__ creates the object, while __init__ initializes the object.

class Person:

    def __new__(cls, name):
        print("__new__ called")
        obj = super().__new__(cls)
        print(type(obj))
        return obj

    def __init__(self, name):
        print("__init__ called")
        self.name = name

p = Person("Suganth")
```

```python
## Modifying the class behaviour using __new__ by over =rriding the default

# Singleton class

class Person:

    _instance = None

    def __new__(cls):

        if cls._instance is None:
            cls._instance = super().__new__(cls)

        return cls._instance

p = Person()
p1 = Person()

p is p1
```

```python
## Descriptors

# A descriptor is a seperate python object that implements descriptor protocols like 
# __get__, __set__ and __delete__. It controls what happens when another object's 
# attribute is read, assigned or deleted.

class Descriptor:

    def __set_name__(self, owner, attribute): #(self, Person, attribute)
        print(f"__init__ {owner} {attribute}")
        self.attribute = attribute

    def __get__(self, instance, owner): #(self, obj, Person)
        print("Getting")
        return instance.__dict__.get(self.attribute)

    def __set__(self, instance, value): #(self, obj, attribute value)
        print("setting")
        instance.__dict__[self.attribute] = value

    def __delete__(self, instance):
        print("deleting")
        del instance.__dict__[self.attribute]

class Person:
    name = Descriptor() # calls __set_name__
    age = Descriptor()
    
```

```python
p = Person() 
p.name = "Suganth" # calls __set__
p.age = 26 # call __set__
p.status = "Single" # directly assigned with descriptors

print(p.name)
print(p.age)
print(p.status)

print(vars(p))

del p.age

print(vars(p))

```
