Q: Why are transformers better than RNNs?

Answer:

* Parallel processing
* Better long context understanding
* Handles dependencies better
* Faster training

Q: Why do LLMs use tokens instead of words?

Answer:

Because tokenization handles:

* Multiple languages
* Unknown words
* Better compression
* Efficient vocabulary management

Q: What is embedding?

Answer:

* Embedding converts text into numerical vector representations where semantically similar texts are located close together.

Q: What is Context Window?

Answer:

* Context window = amount of text model can remember in one request.
* 8K context window - 8000 tokens, includes input prompt, convo history, ouput tokens

        Prompt = 2000 tokens
        History = 3000 tokens
        Output allowed = 3000 tokens

Q: What happens when prompt exceeds context window?

Answer:

* Model truncates old tokens or rejects request depending on implementation.



Q: What is memory in AI agent and why it is needed?

Memory is not something the LLM inherently possesses. The model is stateless. In agentic systems, we implement memory externally. Short-term memory maintains the current conversation context, while long-term memory stores user preferences or historical information in databases or vector stores. During each request, the agent retrieves relevant memory and includes it in the prompt so the LLM can generate context-aware responses.
