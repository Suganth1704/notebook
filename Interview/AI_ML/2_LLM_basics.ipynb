#### What is a Transformer?

A Transformer is a deep learning architecture introduced in the paper Attention Is All You Need that processes text using attention mechanisms instead of sequential processing. It allows parallel computation, understands relationships between words efficiently, and is the foundation of modern LLMs like GPT and Gemini.

Transformer = Attention-based architecture that processes words in parallel and understands context efficiently.

#### Why is it called “Transformer”?

Because it transforms input text into contextual understanding and then predicts the next most likely output.

3. Positional Encoding

        Transformer reads words in parallel.

        So it needs to know word order.

        Example:

        "Dog bites man" is different from "Man bites dog"

        <b>Position encoding tells the model where each word occurs.</b>

4. Self-Attention (Most Important)

        Each word looks at other words to understand meaning.

        Example:

        The cat sat on the mat because it was tired.

        To understand “it”, model checks previous words.

        It learns:

        it → cat

        High-level formula:

        Attention(Q,K,V) = softmax(QKᵀ/√d)V

                * Q = Query
                * K = Key
                * V = Value
        High level:

                * Query asks what to focus on
                * Key tells what information exists
                * Value gives actual information

        Example:

                input: I went to the bank to deposit money.

                Word: bank

                Possible meanings: River bank or Financial bank

                Transformer looks at: Transformer looks at:
        
                Understands: bank = financial institution

5. Multi-Head Attention

        Instead of looking at one relationship, model looks at multiple relationships simultaneously.

        Example:

        Sentence:

        She gave the book to him because he asked.

        Different attention heads may focus on:

                * Header 1 - Grammar
                * Header 2 - Meaning
                * Header 3 - Subject-object relation
                * Header 4 - Pronoun reference

6. Feed Forward Network

        After attention, data passes through neural network layers for further learning.

        Purpose:

                * Learn complex patterns
                * Transform learned representation

7. Output Prediction

        Model predicts next token.

        Example:

        Input:

        I love playing

        Prediction:

        football

        or

        cricket

        depending on probability.

#### What is Temperature?

Temperature in an LLM is a parameter that controls how random or creative the model’s output will be when choosing the next word (token).

In simple terms:

* Low temperature = safer, predictable answers
* High temperature = more creative, varied answers

#### Why Temperature Exists

When models like OpenAI GPT generate text, they don’t “know” one correct answer.

They calculate probabilities for the next token.

Example prompt:

    The capital of France is ___

Possible predictions:

* Paris → 90%
* London → 5%
* Berlin → 3%
* Rome → 2%

Temperature changes how strictly the model follows these probabilities.

Range:

* 0 → deterministic
* 1 → balanced
* 2 → highly creative
