1) let's say we are prompting a LLM but it reaches it maximum token (input token + ouput token) how will be handle this?

##### Uderstanding context window:Every LLM has a context window, which is the maximum number of tokens it can process in a single request.

For example, suppose an LLM has a 128K token context window.


    Input Prompt        : 100K tokens
    Conversation History: 20K tokens
    Expected Output     : 10K tokens
    -------------------------------
    Total               : 130K tokens ❌

*   The request exceeds the context window, so you'll need to handle it <b>before sending it to the model</b>.

<b>Here are the most common strategies.</b>

1) Tuncate (Simple):
    * Remove old context.
    * Instead of sending all 100 messages, keep only the most recent 10–20.

2) Summarize Previous Context
    * Insted of deleting the old messages, summarize then at some point .
    * This preserves important information while using far fewer tokens. Context won't be missed.

        Summary
            +
        Recent Messages
            +
        Current Question
3) Token Counting Before Sending
    * estimate the number of tokens before calling the LLM.

    if token_count(prompt) > MAX_CONTEXT:
        prompt = summarize(prompt)
