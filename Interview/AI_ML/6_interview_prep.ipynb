#### 1) Why cosine similarity ?
- Cosine similarity is used because it directly measures the angular alignment between two vectors. It gives a bounded and intuitive score from -1 to 1, where 1 means the vectors point in the same direction.

- cos 0 = 1 (very similar direcion), cos 90 = 0 (perendicular), cos 180 = -1 (opposite direcion)

            Cosine similarity(A,B) = A.B / ||A||||B||

#### Why not sin and tan?
- sin 0 = 0, sin 90 = 1, sin 180 = 0 
- sine can't distinguish same direction from opposite direction Cosine can

- tan = sin/cos
- tan 90 = undefined (sin 90 = 1, cos 90 = 0)

#### MMR - Maximum Marginal Relevancy

MMR is a retrieval strategy used in RAG to select documents that are both:

- Relevant to the user's query
- Different from each other

- Formula

        MMR=λ×Similarity(D,Q)−(1−λ)×maxSimilarity(D,Dselected​)

Where:

        D = candidate document
        Q = query
        Dselected = documents already selected
        λ = balance between relevance and diversity
