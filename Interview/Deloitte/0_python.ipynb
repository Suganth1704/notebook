```python
'''
Closures in Python

A closure is a function that remembers and can access variable from it enclosing
function even after the enclosing function has finished executing with help of __closure__ attribute


'''
#Example

def outer(x,z):
    def inner(y):
        return f"{x*y} {z}"
    return inner

"""
Outer function returns the inner function with __clousre__ with cell objects
stored in it whihc contains the variable values of outer(enclosing) fn
"""

func = outer(2, z ="dummy")
print(func)
print(func(3))
print(func.__closure__)
print(func.__closure__[0].cell_contents, func.__closure__[1].cell_contents)

func = outer(3, z ="dummy")
print(func)
print(func(3))
print(func.__closure__[0].cell_contents)

## paremeterized decorators are closures

```

```python
'''
Decorators

- Decorators are higher order functions which takes the function as an input/args
retunrs the function as output.

- In other words decorators takes function as input put and modfies it's behaviour.
'''

def decorate(func):
    def wrapper(*args, **kwargs):
        res = list(map(lambda x: x*2, func(*args, **kwargs)))
        return res
    return wrapper

@decorate
def return_list(n):
    return list(range(n))

return_list(4)
```

```python
## Singleton class using decorators

def singleton(cls):
    instance = {}
    def wrapper(*args, **kwargs):
        if instance.get(cls, None) is None:
            instance[cls] = cls(*args, **kwargs)
        return instance[cls]

    return wrapper

@singleton
class Student:
    pass

s1 = Student()
s2 = Student()

print(s1 is s2)
print(f"{id(s1)} - {id(s2)}")
s1.__class__.__annotations__
```

```python
'''
LEGB describes the order Python follows when looking for a variable name:

L → Local → E → Enclosing → G → Global → B → Built-in

'''
x = 10 # global
def outer():
    x = 20 # enclosing

    def inner():
        print(x) # local
    inner()

outer()
```

```python
x = 10 # global
def outer():
    x = 20 # enclosing

    def inner():
        print(x) # local 
        x = 10 ## UnboundLocalError
    inner()

'''
Python sees an assignment to x anywhere inside inner(), 
Python determines that x is a local variable of inner()
'''

outer()
```

```python
x = 10 # global
def outer():
    x = 20 # enclosing

    def inner():
        nonlocal x # global x
        print(x) # local 
        x = 30 ## UnboundLocalError
    inner()

outer()

'''
One important addition:

- global → tells Python to use/modify the global variable.
- nonlocal → tells Python to use/modify a variable from the enclosing scope.
'''
```
