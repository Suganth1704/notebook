| Strategy       | LangChain class                  | Package                                            |
| -------------- | -------------------------------- | -------------------------------------------------- |
| **Fixed-size** | `CharacterTextSplitter`          | `langchain-text-splitters`                         |

| **Structured** | Depends on document type         | `langchain-community` / `langchain-text-splitters` |
| **Semantic**   | `SemanticChunker`                | `langchain-experimental`                           |


```python
import os
from pathlib import Path
BASE_DIR = os.path.dirname(os.path.abspath(__vsc_ipynb_file__))
PARENT_DIR = os.path.dirname(os.path.dirname(BASE_DIR))
TXT_FILE = os.path.join(PARENT_DIR, "data","HTML", "text.txt")

text = Path(TXT_FILE).read_text(encoding="utf-8")
```

```python
## Fixed - size chunking
from langchain_text_splitters import CharacterTextSplitter

splitter = CharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunk = splitter.split_text(text)

#Behaviour is not as expected
```

```python
for i, c in enumerate(chunk):
    print("*"*100)
    print(f"chunk {i}")
    print(c)
```

```python
## Recursive chunking

from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
    separators=["\n\n", "\n", ". ", " ", ""]
)

chunks = splitter.split_text(text)
```

```python
for i, c in enumerate(chunk):
    print("*"*100)
    print(f"chunk {i}")
    print(c)
```
