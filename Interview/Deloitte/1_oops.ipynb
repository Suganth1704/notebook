```python
"""
class vs instance

- A class is a definition/template that describes what objects created
from it can have and do.

It can contain:
- Instance method
- static method
- class attributes
- Descriptors
- __new__() and __init__() for object creation/initialization
"""

#create a class with private variable, property, cls method, statc method, 
# private instance and class variable

class InvalidAge(Exception):
    pass

class Student:

    __school = 'Maria Rafols'

    def __init__(self, name, age):
        super().__init__()
        self.name = name
        self.__age = age

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, age):
        if age <= 0:
            raise InvalidAge
        self.__age = age

    @age.deleter
    def age(self):
        del self.__age
    
    @classmethod
    def creat_obj(cls, name, age):
        return cls(name, age)

    def get_id(self):
        return hash(f"{self.name}")

    def __repr__(self):
        print('__repr__')
        return f"{self.__class__.__name__}({self.name},{self.__age})" 
    
    def __str__(self):
        print('__str__')
        return f"{self.__class__.__name__}({self.name},{self.__age})"    
```

```python
'''
- Composition
Composition is an object-oriented design technique where a class contains 
an instance of another class and uses it to provide functionality. 
It represents a "has-a" relationship and promotes code reuse without 
requiring inheritance.

- Dunder methods

Dunder methods are special methods in python that defines how object behave with built-in
operations and language syntax.

For example, __init__ initializes an object, __str__ defines its string representation,
 __len__ enables len(), and __add__ defines the behavior of the + operator.
'''
```

```python
"""
- Descriptor

Descriptors are Python objects that implement the descriptor protocol like 
__get__, __set__, and __delete__, with which it controls the read, set, and 
deletion operations of the attributes of another object.
"""

class Descriptor:

    def __set_name__(self, owner, attribute):
        print(f"Owner {owner} , attribute {attribute} ")
        self.attribute = attribute

    def __get__(self, instance, owner):
        print('getting')
        return instance.__dict__.get(self.attribute)

    def __set__(self, instance, value):
        print('setting')
        instance.__dict__[self.attribute] = value

    def __delete__(self, instance):
        print('deleting')
        del instance.__dict__[self.attribute]
    
```

```python
class Student:
    name = Descriptor()
    age = Descriptor()

s = Student()
s.name = 'Suganth'
s.age = 26
print(s.name)
print(s.age)
del s.age
print(s.name)
print(s.age)
```

```python
## Class a decorator

class Decorator:

    def __init__(self, func):
        self.func = func
        self.count = 0

    def __call__(self, *args, **kwargs):
        self.count += 1
        print(f'{self.func.__name__} call {self.count}')
        return self.func(*args, **kwargs)

@Decorator
def add(x,y):
    return x+y
```
