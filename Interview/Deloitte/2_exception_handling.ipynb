```python
'''
Blocks in exception handling

- Try
- except
- else - when thereis no exception
- finally - irrespective of error, finally block executes.

Note: even if there is a return in except block, if finally block exist it won't come out of 
function.
'''
```

```python
"""
Exception handling flow
"""
def div(x,y):
    try:
        num = x/y
        
    except Exception as e:
        print(f'Error : {str(e)}')
        return str(e)
    else:
        print('Executed sucessfully')
        return num
    finally:
        print('Clean up')
        #return "cleaned" 
```

```python
div(1,2)
```

```python
div(1,0)
```

```python
## Custom exception

class InvalidAge(Exception):

    def __init__(self, age):
        self.age = age
        # super().__init__(
        #     f"Age {age} is invalid please enter value  > 0"
        # )
        Exception.__init__(self, f"Age {age} is invalid please enter value  > 0")

#remeber Exception(f"Age {age} is invalid please enter value  > 0")

def validate(age):
    if age <= 0:
        raise InvalidAge(age)
    print("Good to go!!")

try:
    validate(0)
except InvalidAge as e:
    print(f"Error : {e}")
```
