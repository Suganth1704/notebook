```python
'''
iterators are the python object that allows you to retrieve elements one at a time.
It implements iterator protocols __iter__ and __next__
'''
#Custom iterator

class Double_counter:

    def __init__(self, limit):
        self.count = 0
        self.limit = limit

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.limit:
            raise StopIteration(f'Reached the limit : {self.limit}')
        self.count += 2
        return self.count
```

```python
d = Double_counter(10)
d.count
print(next(d))
print(next(d))
print(next(d))
print(next(d))
print(next(d))
print(next(d))
print(next(d))
```

```python
'''
A generator is a special type of iterator that is typically created using a 
function containing yield. It produces values lazily, one at a time, 
each time next() is called.
'''

def gen():
    for i in  range(0,2):
        yield i

g = gen()

print(next(g))
g.__next__()


```

```python
## lest yeild chuks

def gen_chunks(chunk_size=10):
    li = list(range(10000))
    for i in range(0,len(li)+1, chunk_size):
        yield li[i:i+chunk_size]

g = gen_chunks()

print(next(g))
print(next(g))
```

```python
'''
yield from

yield from is used inside a generator to delegate yielding 
values to another iterable or generator.
'''

def even_numbers():
    yield 2
    yield 4
    yield 6


def numbers():
    yield 1
    yield from even_numbers()
    yield 7
    yield from [8,9,10]
num = numbers()
print(num.__next__())
print(num.__next__())
print(num.__next__())
print(num.__next__())
print(num.__next__())
print(num.__next__())
```
