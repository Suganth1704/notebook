```python
# Reverse str

st = "Suganth"

print(st[::-1])

def rev_str(st):
    re = ''
    for i in st:
        re = i + re
    return re


print(rev_str(st))
```

```python
# Rev list

def rev_list(li):

    s = 0
    e = len(li)-1
    while s < e:
        li[s], li[e] = li[e], li[s]
        s += 1
        e -=1
    return li

li = [1,2,3,4,5]
print(rev_list(li))
```

```python
## frquency counter

li = [1,2,3,3,4,5]

def freq_count(li):
    d = {}
    for i in li:
        d[i] = d.get(i, 0)+1
    return d

freq_count(li)
```

```python
def fib(n):

    fib = [0,1]
    if n ==0:
        return 0
    else:
        for _ in range(2,n):
            fib.append(fib[-1]+fib[-2])
        return fib
```

```python
fib(7)
```

```python
"""print from 99 to  0 with out loop"""

def rev_seq(num:int) -> int:
    if num != 0:
        print(num)
        return rev_seq(num-1)
    return 0
print(rev_seq(10))
```

```python
def rev_seq(num:int, end:int) -> int:
    if num == end:
        return end
    print(num)
    rev_seq(num+1, end)

rev_seq(0, 10)
```

```python
## bubbel sort

def bubble_sort(li):
    for i in range(len(li)):
        for j in range(0, len(li)-i-1):
            if li[j]> li[j+1]:
                li[j], li[j+1] = li[j+1], li[j]
    return li
```

```python
li = [9,5,6,4,5,8]
bubble_sort(li)
```
