```python
"""
Multiprocessing

- Each process gets its own interpreter and memory.
- Proceess may have multipe threads running on it

"""

from multiprocessing import Process
import time

def task(name):
    print(f"{name} started")
    time.sleep(2)
    print(f"{name} completed")


p1 = Process(target=task, args=("Task 1",))
p2 = Process(target=task, args=("Task 2",))
p1.start()
p2.start()
p1.join()
p2.join()
print("Main process completed")
```
