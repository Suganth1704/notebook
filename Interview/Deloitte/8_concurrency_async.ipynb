## Async - Programming

```python
"""
- coroutine

A coroutine is a python object which can suspend the execution at 'await' and resume it later.
In the meantime the event loop can execute the other task.
"""

"""
- Event loops

The event loop is responsible for scheduling and running coroutines. When a 
coroutine is suspended at an await the event loop can run another task and
later resume the suspended coroutine when its awaited operation is complete
"""
```

```python
"""
A function which is defined with async def is coroutine fucntion.
Calling that function retunrs a coroutine object which can be awaited or
scheduled as an asyncio task
"""
async def fetch_data():
    pass


f = await fetch_data()
print(f)

print(fetch_data())

"""
async - defines the corroutine
await - suspends the current coroutine while wating for an awaitable
"""
```

```python
import asyncio

async def api_1():
    print("API 1 started")
    print('next line is await going to suspend api_1')
    await asyncio.sleep(3)
    print('got respone after await, api_1 resuming the execution')
    print("API 1 finished")

async def api_2():
    print("API 2 started")
    print('next line is await going to suspend api_2')
    await asyncio.sleep(3)
    print('got respone after await, api_2 resuming the execution')
    print("API 2 finished")


async def main():
    await asyncio.gather(
        api_1(),
        api_2()
    )

#asyncio.run(main())

await main()


"""
- API 1 is called, reaches await asyncio.sleep(3), 
- API 1 coroutine is supended. it yields control back to theevent loop 
- and it looks for the another coroutine and finds API2 and excutes 
- API 2 it also encounter the await and it suspends the API 2 coroutine. 
- After 3 seconds the API1 will comes back from sleep and resumes it execution.
"""
```

```python
""""
- Task

Task is an asyncio object that wraps a coroutine and 
schedules it to run on the event loop.

- Future

A Future is an asyncio object that represents a result that 
will be available sometime in the future.
"""


import asyncio

async def api():
    await asyncio.sleep(3)
    return "API response"


async def main():

    task = asyncio.create_task(api())

    print("Task created")
    print('do something')

    result = await task

    print(result)


#asyncio.run(main())
await main()
```

```python
# asyncio.gather() allows multiple awaitable operations to run concurrently and waits for all of them to complete, 
# returning their results in the same order as the awaitables passed to it.

"""
async + await + Task + Future + asyncio → managed by the asyncio event loop.

event loop  - mechanism that runs and schedules asynchronous tasks/coroutines.
"""
```
