```python
"""
replace() vs astype()
"""
df = pd.DataFrame({
    "name": ["A", "B", "B", "C"],
    "age": ["25", "30", "30", "28"],
    "dept": ["IT", "CSE", "ECE", "MECH"]
})
df.info()
```

```python
# astype() -> changes the data type

df['age']=df['age'].astype('int')
```

```python
df.info()
```

```python
df['dept']=df['dept'].replace({
    'IT': "Information technology",
    'CSE': "Computer science",
    'ECE': "Electronic and communication"

})
df
```

```python
df['dept'].rename('Department')
```
