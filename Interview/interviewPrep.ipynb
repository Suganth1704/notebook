```python
'''Given a positive integer n,
count the no. of digits in n that divide n
evenly (i.e., withou leaving a remainder).
return the total no. of such digits'''

def evenlyDivde(n:int, /) -> int:
    count = 0
    num = n
    seen = set()
    while n>0:
        digit = n%10
        if digit !=0 and digit not in seen:
            if num%digit == 0:
                count += 1
            seen.add(digit)
        n//=10
    return count

# 1012%10 = 2 , 1012//10 = 101  
        
```

```python
# Binary search
def binary_search(arr:list, num:int) -> int:
    left, right = 0, len(arr)-1
    while True:
        mid = (left+right) // 2
        if arr[mid] == num:
            return mid
        elif arr[mid] > num:
            right = mid - 1
            
        else:
            left = mid + 1
```

```python
"""
Now let's test whether you can handle strings + frequency/counting without using extra sorting.

Problem

Given a string, find the first non-repeating character.

Return None if every character repeats.

s = "swiss"

Output:
"w"

s = "aabbcc"

Output:
None
"""

def first_non_repeating(s: str) -> str | None:
    freq = {}
    for i in s:
        freq[i] = freq.get(i,0)+1
    
    for k, v in freq.items():
        if v == 1:
            return k
    return None


s = "aabbcc"
print(first_non_repeating(s))
```

```python
"""
this one tests whether you can move beyond simple frequency dictionaries.

Problem: Longest substring without repeating characters

Given a string, find the length of the longest substring that contains no repeated characters.
"""
def longest_unique_substring(s: str) -> int:

    last_seen = {}
    left = 0
    max_length = 0

    for right, char in enumerate(s):
        if char in last_seen:
            left = max(left, last_seen[char]+1)

        last_seen[char] = right

        max_length = max(max_length, right-left +1)
    return max_length

s = "abcabcbb"
longest_unique_substring(s)
```

```python
"""
Now let's test a different skill: stack-based problem solving.
Given a string containing only:

()[]{}

"()[]{}"       → True
"([{}])"       → True
"(]"           → False
"([)]"         → False
"((("          → False
"{[]}"         → True
""             → True

"""

def is_valid(s:str) -> bool:

    stack = []
    pairs = {
        ')': '(',
        ']': '[',
        '}': '{'
    }

    for char in s:
        if char in '([{':
            stack.append(char)
        else:
            if not stack or stack[-1] != pairs[char]:
                return False

            stack.pop()
    return len(stack) == 0
```
