```python
## Count total no. of repeated and non repeated elements
def repeate_non_repeat(*,arr:list) ->tuple:
    # list to dict which tells the frequency
    d = {}
    r_count = 0
    nr_count = 0
    for i in arr:
        d[i] = d.get(i,0)+1 
    # from collections import Counter
    # d = dict(Counter(arr))
    
    for key, val in d.items():
        if val == 1:
            nr_count +=1
        else:
            r_count += 1
    return r_count, nr_count 
```

```python
'''Given a positive integer n,
count the no. of digits in n that divide n
evenly (i.e., withou leaving a remainder).
return the total no. of such digits'''

def evenlyDivde(n:int, /) -> int:
    count = 0
    num = n
    seen = set()
    while n>0:
        digit = n%10
        if digit !=0 and digit not in seen:
            if num%digit == 0:
                count += 1
            seen.add(digit)
        n//=10
    return count

# 1012%10 = 2 , 1012//10 = 101  
        
```

```python
'''Given a list of integers, the gole is to find
the smallest positive integer missing from the list
'''
def missing_small_int(arr:list) -> int:
    arr = list(filter(lambda x: x > 0, arr))
    mx = max(arr)
    sum_n = mx*(mx+1)/2
    su = sum(arr)
    missing_num = sum_n - su
    return missing_num
```

```python
missing_small_int(arr = [1,3,-1,4])
```

```python
## two sum

def two_sum(arr:list, target:int) -> tuple:
    nums = {}
    for i,num in enumerate(arr):
        diff = target - num
        if diff in nums:
            return (nums[diff], i)
        nums[num] = i
    return ()
```

```python
two_sum(arr=[1,2,3,4], target=5)
```

```python
''' The function def differenceofSum(n,m) accepts two
integers n,m as arguments. Find the sum of all numbers in range from 1 to m(both inclusive)
that are not divible by n. Return differenc b/w sum of integers not divisible by n
with sum if numbers divisible by n.
input
n:4
m:20
Output
90'''

def differenceofSum(n:int,m:int) -> int:
    m_range = list(range(1,m+1))
    dif = sum(list(filter(lambda x: x%4 != 0, m_range))) - sum(list(filter(lambda x: x%4 == 0, m_range)))
    return dif

```

```python
'''
You are given a string where multiple characters are repeated consecutively.
You are supposed to reduce the size of thi string using mathematical logic given as in the example below
Input:
aabbbbeeeeffggg

outpu:
a2b4e4f2g3
'''

def reduceSte(st:str) -> str:
    d = {}
    for i in st:
        d[i] = d.get(i,0)+1
    out_str = ''
    for key, val in d.items():
        out_str += f'{key}{val}'
    return out_str
```

```python
def bubble_sort(arr:list) -> list:
    for i in range(len(arr)):
        for j in range(0, len(arr)-i-1):
            if arr[j] > arr[j+1]:
                arr[j],arr[j+1] = arr[j+1], arr[j]
    return arr
```

```python
bubble_sort(arr=[2,3,5,4,6,7,8])
```

```python
# Binary search
def binary_search(arr:list, num:int) -> int:
    left, right = 0, len(arr)-1
    while True:
        mid = (left+right) // 2
        if arr[mid] == num:
            return mid
        elif arr[mid] > num:
            right = mid - 1
            
        else:
            left = mid + 1
```

```python
binary_search(arr=[1,2,3,7,8,9], num=2)
```

```python
"""Return true if duplicate in list"""

def checkDup(arr:list) -> bool:
    d ={}
    for i in arr:
        d[i] = d.get(i,0)+1
    return True if sum(d.values()) == len(d) else False

checkDup([1,2,3,4,5,5])
```
